````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["Salary"], bins=20)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Apply a nice theme
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
````

````python
plt.figure(figsize=(10,6))

sns.histplot(df["Salary"], bins=25, kde=True)

plt.title("Salary Distribution", fontsize=18, fontweight="bold")
plt.xlabel("Salary")
plt.ylabel("Employees")

plt.show()
````

````python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Salary",
    y="Bonus %",
    hue="Gender",
    size="Bonus %",
    alpha=0.7
)

plt.title("Salary vs Bonus",
          fontsize=18,
          fontweight="bold")

plt.show()
````

````python
plt.figure(figsize=(12,6))

sns.countplot(data=df,
              x="Team",
              order=df["Team"].value_counts().index)

plt.xticks(rotation=45)

plt.title("Employees by Team",
          fontsize=18,
          fontweight="bold")

plt.show()
````

````python
avg_salary = df.groupby("Team")["Salary"].mean().sort_values()

plt.figure(figsize=(12,6))

sns.barplot(
    x=avg_salary.index,
    y=avg_salary.values
)

plt.xticks(rotation=45)

plt.title("Average Salary by Team",
          fontsize=18,
          fontweight="bold")

plt.show()
````

````python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="Gender",
    y="Salary"
)

plt.title("Salary Distribution by Gender",
          fontsize=18,
          fontweight="bold")

plt.show()
````

````python
plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="Gender",
    y="Salary"
)

plt.title("Salary Spread",
          fontsize=18,
          fontweight="bold")

plt.show()
````

````python
numeric = df.select_dtypes(include="number")

corr = numeric.corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap",
          fontsize=18,
          fontweight="bold")

plt.show()
````
